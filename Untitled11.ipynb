```python
k_range = range(2, 9)
n_rep = 10
n_boot = 100

# Store results
km_results = {}
boot_results = {k: [] for k in k_range}
```

```python
np.random.seed(1234)

# K-means clustering
for k in k_range:
    models = []
    for _ in range(n_rep):
        kmeans = KMeans(n_clusters=k, n_init=1)
        kmeans.fit(data)
        models.append(kmeans)
    km_results[k] = models
```

```python
!pip install scikit-learn
import numpy as np
import pandas as pd
from sklearn.utils import resample  # Import resample from sklearn.utils
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score

for _ in range(n_boot):
    # Create bootstrap sample
    boot_sample = resample(data)
    for k in k_range:
        # Fit to bootstrap sample
        kmeans = KMeans(n_clusters=k, n_init=1)
        kmeans.fit(boot_sample)

        # Compare with original clustering
        orig_labels = km_results[k][0].predict(data)
        boot_labels = kmeans.predict(data)
        ari = adjusted_rand_score(orig_labels, boot_labels)
        boot_results[k].append(ari)
```

```python
plt.figure(figsize=(12, 10))

# A. Plot number of segments vs cluster quality
plt.subplot(2, 2, 1)
inertias = [np.mean([model.inertia_ for model in km_results[k]]) for k in k_range]
plt.plot(k_range, inertias, 'bo-')
plt.xlabel("Number of segments")
plt.ylabel("Inertia")
plt.title("K-means Quality")
```

```python
# B. Plot stability boxplot
plt.subplot(2, 2, 2)
plt.boxplot([boot_results[k] for k in k_range],
            labels=k_range)
plt.xlabel("Number of segments")
plt.ylabel("Adjusted Rand Index")
plt.title("Cluster Stability")
```

```python
# 3. Select k=4 model
k4_model = km_results[4][0]  # Take first run for k=4
k4_labels = k4_model.predict(data)
```

```python
# 4. Mixture modeling
from sklearn.mixture import BayesianGaussianMixture
bgm = BayesianGaussianMixture(n_components=4,
                             n_init=10,
                             random_state=1234)
bgm.fit(data)
```

```python
# Compare cluster assignments
from sklearn.metrics import confusion_matrix
print("Cluster agreement:")
print(confusion_matrix(k4_labels, bgm.predict(data)))
```

```python
print("\nModel likelihoods:")
print(f"K-means inertia: {k4_model.inertia_:.2f}")
print(f"Mixture model lower bound: {bgm.lower_bound_:.2f}")

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics.pairwise import pairwise_distances
from scipy.optimize import linear_sum_assignment # Import the linear_sum_assignment function

stability_scores = []
for i in range(n_boot):
    boot_sample = resample(data)
    kmeans_boot = KMeans(n_clusters=4, n_init=1).fit(boot_sample)

    # Compute distance between original and bootstrap centers
    dist_matrix = pairwise_distances(k4_model.cluster_centers_,
                                   kmeans_boot.cluster_centers_)
    # Find best matching clusters
    row_ind, col_ind = linear_sum_assignment(dist_matrix)
    stability_scores.append(1 - dist_matrix[row_ind, col_ind].mean())

# Plot overall stability
plt.figure(figsize=(8, 6))
plt.boxplot(stability_scores)
plt.ylim(0, 1)
plt.xlabel("Cluster")
plt.ylabel("Stability Score")
plt.title("Cluster Center Stability")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.mixture import BayesianGaussianMixture
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import adjusted_rand_score
from sklearn.utils import resample
from scipy.cluster.hierarchy import linkage, dendrogram
from statsmodels.formula.api import ols
import seaborn as sns
```

```python
# Set random seed
np.random.seed(1234)


results = {}
for k in range(2, 9):
    models = []
    for _ in range(10):  # nrep=10
        bgm = BayesianGaussianMixture(n_components=k,
                                    max_iter=200,
                                    n_init=1)
        bgm.fit(data)
        models.append(bgm)
    results[k] = models
```

```python
best_models = {}
for k, model_list in results.items():
    best_idx = np.argmax([m.lower_bound_ for m in model_list])
    best_models[k] = model_list[best_idx]

# Plot information criteria
aics = [k*(2*df.shape[1]) - 2*m.lower_bound_ for k, m in best_models.items()]
bics = [k*(2*df.shape[1])*np.log(df.shape[0]) - 2*m.lower_bound_ for k, m in best_models.items()]
```

```python
plt.figure(figsize=(10,6))
plt.plot(range(2,9), aics, 'bo-', label='AIC')
plt.plot(range(2,9), bics, 'ro-', label='BIC')
plt.xlabel('Number of components')
plt.ylabel('Value of criterion')
plt.title('Model Selection Criteria')
plt.legend()
plt.show()
```

```python
# 2. Select k=4 model
m4 = best_models[4]

# Compare with K-means clusters
k4_labels = k4_model.predict(data)
m4_labels = m4.predict(data)

from sklearn.metrics import confusion_matrix
print("Cluster agreement:")
print(confusion_matrix(k4_labels, m4_labels))
```

```python
# 3. Refit model initialized with K-means clusters
m4a = BayesianGaussianMixture(n_components=4,
                             max_iter=200,
                             n_init=1)
m4a.fit(data, y=k4_labels)  # Using K-means as initialization

print("\nLog likelihoods:")
print(f"Original mixture: {m4.lower_bound_}")
print(f"K-means initialized: {m4a.lower_bound_}")
```

```python
# 5. Regression analysis
import statsmodels.api as sm

# Convert 'Like' column to numeric using pd.factorize
# This assigns a unique integer to each distinct value in the 'Like' column
df['Like_numeric'] = pd.factorize(df['Like'])[0]

# Update the formula to use the new numeric 'Like_numeric' column
formula = "Like_numeric ~ " + " + ".join(data.columns)

model = sm.OLS.from_formula(formula, data=pd.concat([df['Like_numeric'], data], axis=1))
results = model.fit()
print(results.summary())
```

```python
# 9. Decision tree (using sklearn)
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder  # Import OrdinalEncoder

# Create a LabelEncoder object
le = LabelEncoder()

# Fit the encoder to the 'Like' column and transform it
df['Like_encoded'] = le.fit_transform(df['Like'])

# Create an OrdinalEncoder object for 'VisitFrequency'
ordinal_encoder = OrdinalEncoder()

# Reshape 'VisitFrequency' before encoding
visit_frequency_reshaped = df['VisitFrequency'].values.reshape(-1, 1)

# Fit and transform 'VisitFrequency' using the OrdinalEncoder
df['VisitFrequency_encoded'] = ordinal_encoder.fit_transform(visit_frequency_reshaped)

# Create a LabelEncoder for the 'Gender' column
gender_encoder = LabelEncoder()

# Fit and transform the 'Gender' column
df['Gender_encoded'] = gender_encoder.fit_transform(df['Gender'])


tree = DecisionTreeClassifier(max_depth=3)

# Use the encoded columns in the fit method, including 'Gender_encoded'
tree.fit(df[['Like_encoded', 'Age', 'VisitFrequency_encoded', 'Gender_encoded']],  # Use encoded 'VisitFrequency' and 'Gender'
         k4_labels == 2)  # Cluster 3 in R is index 2 in Python (0-based)

plt.figure(figsize=(15,10))
# Update feature_names to include 'Gender_encoded'
plot_tree(tree, feature_names=['Like_encoded', 'Age', 'VisitFrequency_encoded', 'Gender_encoded'])
plt.show()
```

```python
# 10. Segment profiles
# Calculate the mode (most frequent value) for 'VisitFrequency' instead of the mean
visit_modes = df.groupby(k4_labels)['VisitFrequency'].agg(lambda x: x.mode()[0] if not x.empty else np.nan)

# Extract the numeric part of the 'Like' column using regex
df['Like_numeric'] = df['Like'].str.extract(r'([-+]?\d+)').astype(float)

like_means = df.groupby(k4_labels)['Like_numeric'].mean()
female_means = (df['Gender'] == 'Female').groupby(k4_labels).mean()

plt.figure(figsize=(8,6))

plt.scatter(visit_modes, like_means, s=female_means*1000, alpha=0.6)
for i in range(4):
    plt.text(visit_modes[i], like_means[i], str(i+1), ha='center', va='center')
plt.xlabel('Most Frequent Visit Frequency') # Update x-axis label
plt.ylabel('Average Like Score')
plt.title('Segment Profiles (Size = % Female)')
plt.grid(True)
plt.show()
```
